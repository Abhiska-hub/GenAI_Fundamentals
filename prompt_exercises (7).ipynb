## Part A - Zero-Shot vs. Few-Shot Prompting (Factual Verification)

**Claim to verify:** *"The Pythagorean theorem was already described in Bhartiya (Indian) Vedic scriptures — before Pythagoras."*

**Exercise A1.** Ask the model to fact-check this claim using a **zero-shot** prompt (just ask directly, no examples of how to structure a fact-check). Record the verdict and whether sources/evidence are cited.


```python
import os
from huggingface_hub import InferenceClient

# Initialize client using your environment token
client = InferenceClient(api_key=os.getenv("HF_TOKEN"))

def ask_hf(prompt, max_tokens=300):
    response = client.chat_completion(
        model="Qwen/Qwen2.5-72B-Instruct",
        messages=prompt,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content
```

```python
#Exercise A1
zero_shot_prompt = [
    {
        "role": "user",
        "content": "Fact-check this claim: 'The Pythagorean theorem was already described in Bhartiya (Indian) Vedic scriptures — before Pythagoras.' Provide a clear verdict and cite evidence or sources."
    }
]

output_a1 = ask_hf(prompt=zero_shot_prompt, max_tokens=300)
print(output_a1)
```

**Exercise A2.** Repeat the same fact-check using a **few-shot** prompt — first give the model 2–3 example claims already fact-checked in a template format (Claim → Verdict → Evidence → Caveats), then ask it to fact-check the Pythagorean theorem claim in the same format.


```python
#Exercise A2
few_shot_prompt = [
    {
        "role": "user",
        "content": """Please fact-check claims using the following exact format:
Claim: [Statement]
Verdict: [True / False / Partially True]
Evidence: [Historical or scientific evidence with sources]
Caveats: [Important context, or limitations]

---
Example 1:
Claim: The concept of zero as a number was invented in ancient India.
Verdict: True
Evidence: Ancient Indian mathematician Brahmagupta defined zero and its algebraic rules in his work Brahmasphutasiddhanta (628 CE). Earlier references to zero as a place-value marker exist in texts like the Bakhshali manuscript.
Caveats: While India formalized zero as a mathematical number with operations, placeholder concepts existed in ancient Sumerian and Mayan cultures independently.

Example 2:
Claim: "Humans only use 10 percentage of their brain."
Verdict: False
Evidence: Neuroimaging studies show that almost all parts of the brain are active during daily tasks, even during sleep.
Caveats: While we don't activate every single neuron simultaneously (which would cause a seizure), 100 percentage of the brain has known functions.
---

Now fact-check this claim using the exact same format:
Claim: 'The Pythagorean theorem was already described in Bhartiya (Indian) Vedic scriptures — before Pythagoras.'"""
    }
]

output_a2 = ask_hf(prompt=few_shot_prompt, max_tokens=400)
print(output_a2)
```

**Reflection:** Did the few-shot version produce a more structured, nuanced, or better-sourced answer than the zero-shot version? Did either version oversimplify the claim (e.g., stating it as flatly true/false when the historical picture is more nuanced)?

Answer:
Structure: Zero-shot gave a plain paragraph. Few-shot forced a clean format (Verdict -> Evidence -> Caveats).
Nuance: Few-shot handled details much better
oversimplify:Zero-shot gives a True/False, while few-shot breaks down the line between practical ancient use and  mathematical proof.


## Part B - Prompt Structure & Specificity

**Exercise B1.** Write a **vague** prompt asking the model to "write something about climate change." Record the output.



```python
#Exercise B1
vague_prompt = [{"role": "user", "content": "Write something about climate change."}]

output_b1 = ask_hf(vague_prompt)
print(output_b1)
```

**Exercise B2.** Rewrite the same request as a **specific, well-structured prompt** that includes: role/persona, target audience, desired length, tone, and output format (e.g., bullet points vs. paragraph).



```python
# Exercise B2 — Simple Structured Prompt
structured_prompt = [
    {
        "role": "user",
        "content": "Act as a high school teacher. Explain climate change causes and solutions in 3 short bullet points using simple words."
    }
]

output_b2 = ask_hf(prompt=structured_prompt, max_tokens=150)
print(output_b2)
```

**Reflection:** List 3 concrete differences between the two outputs (length, tone, usefulness, structure).

Structure: B1 produces a dense, generic paragraph; B2 uses clean bullet points that are easy to understand.

Tone: B1 sounds like a dry content while B2 uses simple, language.

Usefulness: B1 gives a broad overview while B2 gives a  focused summary.

## Part C - Chain-of-Thought Prompting

**Exercise C1.** Give the model a multi-step reasoning problem (e.g., a word problem or logic puzzle) with a plain, direct prompt. Record whether the answer is correct.




```python
# Exercise C1 — Direct Prompting
direct_prompt = [
    {
        "role": "user",
        "content": "A store had 45 laptops. They sold 1/3 of them on Monday and 8 more on Tuesday. On Wednesday, they received a shipment that doubled their current stock. How many laptops do they have now? Give only the final answer."
    }
]

output_c1 = ask_hf(prompt=direct_prompt, max_tokens=100)
print(output_c1)
```

**Exercise C2.** Ask the same question again, this time adding "Let's think step by step" (or explicitly asking it to show its reasoning before the final answer).



```python
# Exercise C2 — Chain-of-Thought (CoT) Prompting
cot_prompt = [
    {
        "role": "user",
        "content": "A store had 45 laptops. They sold 1/3 of them on Monday and 8 more on Tuesday. On Wednesday, they received a shipment that doubled their current stock. How many laptops do they have now? Let's think step by step before giving the final answer."
    }
]

output_c2 = ask_hf(prompt=cot_prompt, max_tokens=300)
print(output_c2)
```

**Reflection:** Did explicitly requesting reasoning improve correctness or clarity? Note any cases where showing steps didn't help.
